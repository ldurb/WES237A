```python
led4=rgbled.RGBLED(4)
led5=rgbled.RGBLED(5)

for i in range(10): 
    for x in range(0,7):
        led4.write(x)
        led5.write(x+1)
        time.sleep(1)

```

```python
import pynq.lib.rgbled as rgbled
import time
led4=rgbled.RGBLED(4)
led5=rgbled.RGBLED(5)

red = 0x1
blue = 0x4
for i in range(100): #police
    led4.write(red)
    led5.write(blue)
    red,blue=blue,red
    time.sleep(.1)
```
